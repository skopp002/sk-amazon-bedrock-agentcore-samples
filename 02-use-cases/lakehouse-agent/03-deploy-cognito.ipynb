## Step 1: Run Cognito Setup

```python
import subprocess

result = subprocess.run(
    ['python', 'setup_cognito.py', '--region', config.AWS_REGION],
    cwd='gateway-setup',
    capture_output=True,
    text=True
)

print(result.stdout)
if result.returncode != 0:
    print('❌ Error:', result.stderr)
else:
    print('\n✅ Cognito setup complete!')
    print('\n📋 Save these values in the next step')
```

## Step 2: Save Cognito Configuration to SSM

**IMPORTANT**: Copy the values from Step 1 output and paste below.

```python
# UPDATE THESE with actual values from Step 1
COGNITO_USER_POOL_ID = 'us-east-1_XXXXX'  # CHANGE THIS
COGNITO_APP_CLIENT_ID = 'your-client-id'  # CHANGE THIS
COGNITO_APP_CLIENT_SECRET = 'your-secret'  # CHANGE THIS
COGNITO_DOMAIN = 'https://your-domain.auth.region.amazoncognito.com'  # CHANGE THIS

# Save to SSM
params = {
    'lh_cognito_user_pool_id': (COGNITO_USER_POOL_ID, 'String'),
    'lh_cognito_app_client_id': (COGNITO_APP_CLIENT_ID, 'String'),
    'lh_cognito_app_client_secret': (COGNITO_APP_CLIENT_SECRET, 'SecureString'),
    'lh_cognito_domain': (COGNITO_DOMAIN, 'String')
}

for name, (value, param_type) in params.items():
    ssm_client.put_parameter(Name=name, Value=value, Type=param_type, Overwrite=True)
    print(f'✅ Saved {name}')

print('\n✅ All Cognito parameters saved to SSM!')
```
