```python
import boto3
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd()))
from config import config

ssm_client = boto3.client('ssm')

print('✅ Setup complete')
print(f'   RLS Role: {config.RLS_ROLE_ARN}')
```

## Step 1: Deploy MCP Server

**Note**: This requires Docker to be running.

```python
import subprocess

result = subprocess.run(
    ['python', 'deploy_runtime.py'],
    cwd='mcp-lakehouse-server',
    capture_output=True,
    text=True
)

print(result.stdout)
if result.returncode != 0:
    print('❌ Error:', result.stderr)
else:
    print('\n✅ MCP Server deployed!')
    print('\n📋 Save the Runtime ARN in the next step')
```

## Step 2: Save MCP Server ARN to SSM

```python
# UPDATE THIS with actual ARN from Step 1
MCP_SERVER_RUNTIME_ARN = 'arn:aws:bedrock-agentcore:region:account:runtime/id'  # CHANGE THIS
MCP_SERVER_RUNTIME_ID = 'runtime-id'  # CHANGE THIS

ssm_client.put_parameter(Name='lh_mcp_server_runtime_arn', Value=MCP_SERVER_RUNTIME_ARN, Type='String', Overwrite=True)
ssm_client.put_parameter(Name='lh_mcp_server_runtime_id', Value=MCP_SERVER_RUNTIME_ID, Type='String', Overwrite=True)

print('✅ Saved MCP Server configuration to SSM')
```
