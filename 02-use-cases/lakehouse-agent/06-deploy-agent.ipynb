## Prerequisites

- ✅ Run `05-deploy-gateway.ipynb` first
- ✅ Gateway ARN saved to SSM
- ✅ Docker installed and running

## What This Notebook Does

1. Deploys Lakehouse Agent to AgentCore Runtime

```python
import boto3
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd()))
from config import config

ssm_client = boto3.client('ssm')

print('✅ Setup complete')
print(f'   Gateway ARN: {config.GATEWAY_ARN}')
```

## Step 1: Deploy Lakehouse Agent

```python
import subprocess

result = subprocess.run(
    ['python', 'deploy_lakehouse_agent.py'],
    cwd='lakehouse-agent',
    capture_output=True,
    text=True
)

print(result.stdout)
if result.returncode != 0:
    print('❌ Error:', result.stderr)
else:
    print('\n✅ Lakehouse Agent deployed!')
    print('\n📋 Configuration automatically saved to SSM')
```

## Step 2: Verify Deployment

```python
# Get saved configuration
AGENT_RUNTIME_ARN = ssm_client.get_parameter(Name='lh_lakehouse_agent_runtime_arn')['Parameter']['Value']
AGENT_RUNTIME_ID = ssm_client.get_parameter(Name='lh_lakehouse_agent_runtime_id')['Parameter']['Value']

print('✅ Agent Configuration:')
print(f'   Runtime ARN: {AGENT_RUNTIME_ARN}')
print(f'   Runtime ID: {AGENT_RUNTIME_ID}')
```
